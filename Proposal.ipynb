## Description of Variables

After running the code, a compressed package will be downloaded, which contains the Python code for data extraction, a file with IDs for Indian regions based on latitude and longitude, and a large amount of weather data sorted alphabetically by region.Here are the description of variables: 

**date**: The date of the recorded weather data.

**temperature_2m**: Air temperature measured at a height of 2 meters above ground level.

**relative_humidity_2m**: Relative humidity measured at a height of 2 meters above ground level, representing the percentage of moisture in the air compared to the maximum it can hold at that temperature.

**dew_point_2m**: Dew point temperature at 2 meters above ground level, indicating the temperature at which air becomes saturated and dew forms.

**apparent_temperature**: The perceived temperature, considering factors such as wind and humidity, often referred to as the "feels-like" temperature.

**precipitation**: The total amount of precipitation (rain, snow, sleet, etc.) that occurred during the measurement period, usually measured in millimeters.

**rain**: The amount of rainfall recorded during the measurement period.

**snowfall**: The amount of snow that fell during the measurement period.

**snow_depth**: The total depth of accumulated snow on the ground at the time of measurement.
